This arrangement is of course is not necessary legal, so we'll write a simple DFS search with backtracking:

First some helper functions ..


1. place_in_next_row

2. remove_from_current_row

3. is_place_in_next_row_valid

```python
def place_in_next_row(col, columns):
    columns.append(col)

def remove_from_current_row(columns):
    if len(columns) > 0:
        return columns.pop()
    return -1

# better version

def is_place_in_next_row_valid(col, columns):
    # new row
    row = len(columns)

    # for each row with existing queen check if the current (row, col) position is not valid

    for queen_row, queen_column in enumerate(columns):
        if col == queen_column:
            return False
          # check if col in new row is in the same diagonal as existing queen
        if queen_column - queen_row == col - row:
            return False
          # check if col in new row is in the same anti-diagonal as existing queen
        if queen_column + queen_row == col + row:
            return False

    return True
```

We can now write the desired dfs code by starting with a blank board and placing the queens in one row at a time, starting with the first row.



```python
def solve_queens_dfs(size):
    columns = []
    number_of_moves = 0
    number_of_iterations = 0
    row = 0 # current row
    col = 0 # always start each row at leftmost col
    # iterate over rows of board
    while True:
        #place queen in next row
        # print("I am trying to put the queen in row ", row)
        # print(columns)
        # print(f"number_of_moves: {number_of_moves}")
        while col < size:
            number_of_iterations += 1
            if is_place_in_next_row_valid(col, columns):
                place_in_next_row(col, columns)
                number_of_moves+=1
                # print(f"placed: row: {row}, col: {col}")
                row += 1
                col = 0
                break
            else:
                # print(f"not placed: row: {row}, col: {col}")
                col += 1

        # could not find an open col in this row,  or board is full
        if (col == size or row == size):
            number_of_iterations+=1
            # if board is full, we have a solution
            if row == size:
                #print("I did it! Here is my solution")
                #display(columns)
                converged = True
                return columns, number_of_iterations, number_of_moves, converged
            # else couldn't find a solution so need to backtrack
            #print("start to backtrack ... ")
            prev_col = remove_from_current_row(columns)
            number_of_moves+=1
            if (prev_col == -1): # backtracked past column 1
                print("There are no solutions")
                #print(number_of_moves)
                converged = False
                return columns, number_of_iterations, number_of_moves, converged
            # retry previous row again
            row -= 1
            # start to now check at col = (1 + value of prev_column in the row)
            col = 1 + prev_col

```

## Shared utilities

Everything below uses the same board representation as the code above: `columns[row] = col`.

* `is_valid_solution` checks every board an algorithm returns, so a "solved" count can never include a wrong board.
* `ConflictBoard` is the main speed-up for all the hill-climbing variants. Instead of comparing a queen against every other queen (O(n)), it keeps three counters: queens per column, queens per `row+col` diagonal and queens per `row-col` diagonal. That makes "how many queens attack square (r, c)?" an O(1) lookup, and moving a queen is an O(1) update of the counters.
* `run_benchmark` times one algorithm over a range of n with `time.time()` (same tic/toc pattern as the timing example from class) and adds up the time and the counters.

**Counters.** Every solver returns `(columns, number_of_iterations, number_of_moves, converged)`, just like `solve_queens_dfs`:
* `number_of_iterations` = how many candidate positions or boards the algorithm *checked*
* `number_of_moves` = how many times a queen was actually placed, removed or moved

Some solvers also fill an optional `stats` dictionary with extra counters (restarts, sideways moves, backtracks).

```python
import time
import statistics

random.seed(42)  # makes the random algorithms reproducible

def is_valid_solution(columns):
    n = len(columns)
    if n == 0 or any(not (0 <= c < n) for c in columns):
        return False
    # no two queens share a column, a row+col diagonal or a row-col diagonal
    return (len(set(columns)) == n
            and len({r + c for r, c in enumerate(columns)}) == n
            and len({r - c for r, c in enumerate(columns)}) == n)


class ConflictBoard:
    """A board that knows, in O(1), how many queens attack any square."""

    def __init__(self, columns):
        n = len(columns)
        self.n = n
        self.columns = list(columns)
        self.col_count = [0] * n
        self.diag_sum = [0] * (2 * n - 1)    # indexed by row + col
        self.diag_diff = [0] * (2 * n - 1)   # indexed by row - col + n - 1
        for r, c in enumerate(self.columns):
            self._add(r, c, 1)

    def _add(self, r, c, delta):
        self.col_count[c] += delta
        self.diag_sum[r + c] += delta
        self.diag_diff[r - c + self.n - 1] += delta

    def attacks_on(self, r, c):
        """Number of queens (other than row r's own queen) attacking square (r, c)."""
        total = self.col_count[c] + self.diag_sum[r + c] + self.diag_diff[r - c + self.n - 1]
        if self.columns[r] == c:
            total -= 3  # don't count the queen itself
        return total

    def queen_conflicts(self, r):
        return self.attacks_on(r, self.columns[r])

    def move(self, r, new_c):
        self._add(r, self.columns[r], -1)
        self.columns[r] = new_c
        self._add(r, new_c, 1)

    def total_conflicts(self):
        return sum(self.queen_conflicts(r) for r in range(self.n)) // 2


def run_benchmark(name, solver, ns, verbose=False):
    """Solve every n in ns, time each one, validate it and add up the counters."""
    total_time = 0.0
    solved = total_iterations = total_moves = 0
    for n in ns:
        tic = time.time()
        columns, iterations, moves, converged = solver(n)
        toc = time.time()
        if converged:
            assert is_valid_solution(columns), f"{name} returned an invalid board for n={n}"
        total_time += toc - tic
        solved += converged
        total_iterations += iterations
        total_moves += moves
        if verbose:
            print(f"  {name:<40} n={n:<3} time={toc - tic:8.4f}s  iterations={iterations:<10} "
                  f"moves={moves:<9} solved={converged}")
    return {"name": name, "time": total_time, "solved": solved, "count": len(ns),
            "iterations": total_iterations, "moves": total_moves}


def report_counters(name, solver, ns=(30, 40, 50), runs_at_100=30):
    """What the homework cells ask for: counters for n=30, 40, 50 and averages of 30 runs at n=100."""
    print(name)
    for n in ns:
        stats = {}
        columns, iterations, moves, converged = solver(n, stats)
        if converged:
            assert is_valid_solution(columns)
        print(f"  n={n}: iterations={iterations}, moves={moves}, solved={converged} {stats if stats else ''}")
    if runs_at_100:
        results = []
        for _ in range(runs_at_100):
            tic = time.time()
            columns, iterations, moves, converged = solver(100, None)
            results.append((iterations, moves, converged, time.time() - tic))
        print(f"  n=100, {runs_at_100} runs: avg iterations={statistics.mean(r[0] for r in results):.0f}, "
              f"avg moves={statistics.mean(r[1] for r in results):.0f}, "
              f"solved {sum(r[2] for r in results)} of {runs_at_100}, "
              f"avg time={statistics.mean(r[3] for r in results):.4f}s")
```

```python
def solve_queens_british_museum(n, max_iterations=None):
    """British Museum: keep putting down random boards until one happens to be legal."""
    number_of_iterations = 0
    number_of_moves = 0
    while max_iterations is None or number_of_iterations < max_iterations:
        columns = place_n_queens(n)
        number_of_iterations += 1   # one whole board generated and checked
        number_of_moves += n        # n queens put down
        if is_valid_solution(columns):
            return columns, number_of_iterations, number_of_moves, True
    return columns, number_of_iterations, number_of_moves, False


for n in (4, 6, 8):
    tic = time.time()
    columns, num_iterations, number_moves, converged = solve_queens_british_museum(n)
    toc = time.time()
    print(f"n={n}: number of iterations: {num_iterations}, number of moves: {number_moves}, "
          f"time: {toc - tic:.3f}s")
    displayBoard(columns)
```

**Why British Museum is faster here than the handout's "4 billion".** The handout's figure is the number of ways to put 8 queens on any 8 of the 64 squares: C(64, 8) ≈ 4.4 billion. `place_n_queens` already puts exactly one queen in each row, so the space is only 8⁸ ≈ 16.7 million boards, and 92 of them are solutions. So we expect about 16.7M / 92 ≈ 182,000 random boards before hitting one, which takes well under a second. n=4 (2 solutions out of 256 boards) and n=6 (4 out of 46,656) are even quicker. Past n=8 the space grows like nⁿ, so we stop there, as the handout says.

### Hill climbing (heuristic repair)

All four variants start from a random board (`place_n_queens`) and repair it using `ConflictBoard`. Counters:
* `number_of_iterations` = squares or queens whose conflicts we *looked at*
* `number_of_moves` = queens actually moved

| Variant | Which queen to fix | Where to move it | When stuck |
|---|---|---|---|
| **A. Basic** | most-conflicted queen (O(n) scan) | its least-conflicted column | give up |
| **B. Restart** | most-conflicted queen (O(n) scan) | its least-conflicted column | restart with a new random board (also after `max_steps`) |
| **C. Sideways** | most-conflicted queen (O(n) scan) | its least-conflicted column, *or* an equally good different column | sideways up to `max_sideways` times, then restart (also after `max_steps`) |
| **D. Any improvement** | a *random* queen that has a conflict (no scan) | the first column, in random order, that improves; otherwise a random equally good one | restart after `max_steps` |

Ties are always broken randomly. Without that, the same queen keeps getting picked and A–C cycle.

The default parameters below were picked using the **Parameter tuning** section further down.

```python
def best_columns_for(board, r):
    """Least-conflicted columns for row r's queen. Looks at all n columns."""
    best, best_cols = None, []
    for c in range(board.n):
        a = board.attacks_on(r, c)
        if best is None or a < best:
            best, best_cols = a, [c]
        elif a == best:
            best_cols.append(c)
    return best, best_cols


def most_constrained_queen(board):
    """The "fancy" check: scan every queen and return the one with the most conflicts."""
    worst, rows = -1, []
    for r in range(board.n):
        q = board.queen_conflicts(r)
        if q > worst:
            worst, rows = q, [r]
        elif q == worst:
            rows.append(r)
    return worst, random.choice(rows)


# ---------- A. Basic hill climbing ----------
def hc_basic(n, stats=None):
    board = ConflictBoard(place_n_queens(n))
    number_of_iterations = number_of_moves = 0
    while True:
        worst, r = most_constrained_queen(board)
        number_of_iterations += n
        if worst == 0:
            return board.columns, number_of_iterations, number_of_moves, True
        best, best_cols = best_columns_for(board, r)
        number_of_iterations += n
        if best >= worst:   # this queen can't be improved, so we're stuck
            return board.columns, number_of_iterations, number_of_moves, False
        board.move(r, random.choice(best_cols))
        number_of_moves += 1


columns, num_iterations, number_moves, converged = hc_basic(8)
print(f"n=8: iterations={num_iterations}, moves={number_moves}, solved={converged}")
displayBoard(columns)
```

```python
# ---------- B and C share one loop: B is C with max_sideways = 0 ----------
def hc_with_restarts(n, max_steps, max_sideways, time_limit=None, stats=None):
    tic = time.time()
    number_of_iterations = number_of_moves = restarts = sideways_moves = 0
    while True:
        board = ConflictBoard(place_n_queens(n))   # step 1: put down n queens
        sideways_left = max_sideways
        for step in range(max_steps):
            worst, r = most_constrained_queen(board)   # step 2
            number_of_iterations += n
            if worst == 0:
                if stats is not None:
                    stats.update(restarts=restarts, sideways=sideways_moves)
                return board.columns, number_of_iterations, number_of_moves, True
            best, best_cols = best_columns_for(board, r)   # step 3
            number_of_iterations += n
            if best < worst:
                board.move(r, random.choice(best_cols))
                number_of_moves += 1
            elif best == worst and sideways_left > 0:      # sideways step (variant C only)
                others = [c for c in best_cols if c != board.columns[r]]
                if not others:
                    break
                board.move(r, random.choice(others))
                number_of_moves += 1
                sideways_left -= 1
                sideways_moves += 1
            else:
                break                                       # stuck, so restart
        if board.total_conflicts() == 0:
            if stats is not None:
                stats.update(restarts=restarts, sideways=sideways_moves)
            return board.columns, number_of_iterations, number_of_moves, True
        restarts += 1
        if time_limit is not None and time.time() - tic > time_limit:
            if stats is not None:
                stats.update(restarts=restarts, sideways=sideways_moves)
            return board.columns, number_of_iterations, number_of_moves, False


# ---------- B. Hill climbing with restart ----------
def hc_restart(n, max_steps=100, time_limit=1.0, stats=None):
    return hc_with_restarts(n, max_steps, 0, time_limit, stats)


# ---------- C. Hill climbing with sideways moves + restart ----------
def hc_sideways(n, max_steps=200, max_sideways=100, time_limit=10.0, stats=None):
    return hc_with_restarts(n, max_steps, max_sideways, time_limit, stats)
```

```python
# ---------- D. Any improvement, random queen (min-conflicts style) ----------
def hc_min_conflicts(n, max_steps=1000, time_limit=10.0, stats=None):
    tic = time.time()
    number_of_iterations = number_of_moves = restarts = 0
    cols_order = list(range(n))
    while True:
        board = ConflictBoard(place_n_queens(n))
        total = board.total_conflicts()
        for step in range(max_steps):
            if total == 0:
                break
            # pick a random queen that is under attack; no scan for the "worst" one
            while True:
                r = random.randrange(n)
                number_of_iterations += 1
                current = board.queen_conflicts(r)
                if current > 0:
                    break
            # try its columns in random order and take the FIRST improvement
            random.shuffle(cols_order)
            chosen, equal_cols = None, []
            for c in cols_order:
                if c == board.columns[r]:
                    continue
                number_of_iterations += 1
                a = board.attacks_on(r, c)
                if a < current:
                    chosen = c
                    break
                if a == current:
                    equal_cols.append(c)
            if chosen is None and equal_cols:
                chosen = equal_cols[0]    # no improvement, so take a random sideways move
            if chosen is not None:
                total += board.attacks_on(r, chosen) - current
                board.move(r, chosen)
                number_of_moves += 1
        if total == 0:
            if stats is not None:
                stats.update(restarts=restarts)
            return board.columns, number_of_iterations, number_of_moves, True
        restarts += 1
        if time_limit is not None and time.time() - tic > time_limit:
            if stats is not None:
                stats.update(restarts=restarts)
            return board.columns, number_of_iterations, number_of_moves, False
```

### Forward checking

Backtracking (like DFS), but every row keeps a **domain**: the set of columns still legal for it.
* **Domains as bitmasks.** Each domain is a Python int where bit `c` set means "column c is still allowed", so removing values is a single `&= ~mask`.
* **Forward checking.** When a queen goes on (r, c), we remove column c and the two diagonal squares from every *unassigned* row's domain. If any domain becomes empty, that placement can't work, and we undo it right away instead of discovering the dead end several rows later. This is the "more sophisticated `is_place_in_next_row_valid`": the check happens once, ahead of time, for all future rows.
* **MRV (minimum remaining values).** Instead of filling rows top to bottom, we always fill the row with the *fewest* legal columns left, the most constrained one. This is what takes it from about 25 queens (plain DFS) to 100. Ties are broken randomly, and the columns are tried in random order, so each run is different, which is why we average 30 runs at n=100.
* Optional `max_backtracks`: give up and restart after that many backtracks. The tuning section shows it isn't needed for n ≤ 100.

Counters: `number_of_iterations` = columns tried, `number_of_moves` = queens placed plus queens removed, `stats["backtracks"]` = placements undone.

```python
def solve_queens_fc(n, max_backtracks=None, time_limit=None, stats=None):
    tic = time.time()
    number_of_iterations = number_of_moves = backtracks = restarts = 0
    while True:
        domains = [(1 << n) - 1] * n    # every column allowed in every row
        columns = [-1] * n
        unassigned = set(range(n))
        bt = 0

        def assign():
            nonlocal number_of_iterations, number_of_moves, bt
            if not unassigned:
                return True
            # MRV: the row with the fewest legal columns left (random tie-break)
            best_size, best_rows = n + 1, []
            for r2 in unassigned:
                size = domains[r2].bit_count()
                if size < best_size:
                    best_size, best_rows = size, [r2]
                elif size == best_size:
                    best_rows.append(r2)
            r = random.choice(best_rows)
            values = [c for c in range(n) if domains[r] >> c & 1]
            random.shuffle(values)
            unassigned.remove(r)
            for c in values:
                number_of_iterations += 1
                # forward check: remove (r, c)'s column and diagonals from every future row
                saved = []
                ok = True
                for r2 in unassigned:
                    d = abs(r2 - r)
                    attacked = 1 << c
                    if c + d < n:
                        attacked |= 1 << (c + d)
                    if c - d >= 0:
                        attacked |= 1 << (c - d)
                    if domains[r2] & attacked:
                        saved.append((r2, domains[r2]))
                        domains[r2] &= ~attacked
                        if domains[r2] == 0:   # a future row has no legal column left
                            ok = False
                            break
                columns[r] = c
                number_of_moves += 1
                if ok and assign():
                    return True
                for r2, dom in saved:          # undo the pruning
                    domains[r2] = dom
                columns[r] = -1
                number_of_moves += 1
                bt += 1
                if max_backtracks is not None and bt > max_backtracks:
                    break
            unassigned.add(r)
            return False

        solved = assign()
        backtracks += bt
        if solved or max_backtracks is None or (time_limit is not None and time.time() - tic > time_limit):
            if stats is not None:
                stats.update(backtracks=backtracks, restarts=restarts)
            return columns, number_of_iterations, number_of_moves, solved
        restarts += 1


columns, num_iterations, number_moves, converged = solve_queens_fc(8)
print(f"n=8: iterations={num_iterations}, moves={number_moves}, solved={converged}")
displayBoard(columns)
report_counters("Forward checking (MRV)", lambda n, s: solve_queens_fc(n, stats=s))
```

### Parameter tuning

Each setting is run on n = 20, 50 and 100 (B: 20, 50 and 80), with several seeds each, and every run is capped at 2 seconds (B uses its own 1-second cap). The output shows the mean time per solve and how many solved within the cap. The setting used as the default in each function is the one picked from these results; the reasoning is in the write-up at the end.

```python
def tune(label, make_solver, settings, ns=(20, 50, 100), seeds=5):
    print(label)
    for setting in settings:
        solver = make_solver(setting)
        total, solved, runs = 0.0, 0, 0
        for n in ns:
            for seed in range(seeds):
                random.seed(seed)
                tic = time.time()
                columns, iterations, moves, converged = solver(n)
                total += time.time() - tic
                solved += converged
                runs += 1
        print(f"  {setting!s:>6}: mean time {total / runs:.4f}s, solved {solved}/{runs}")


# every tuning run gets at most LIMIT seconds; a setting that needs more counts as unsolved
LIMIT = 2.0
tune("D. max_steps before restart",
     lambda s: (lambda n: hc_min_conflicts(n, max_steps=s, time_limit=LIMIT)), [10, 50, 100, 500, 1000, 5000])
tune("C. max_sideways (max_steps=200)",
     lambda s: (lambda n: hc_sideways(n, max_steps=200, max_sideways=s, time_limit=LIMIT)), [0, 10, 50, 100, 200])
tune("C. max_steps (max_sideways=50)",
     lambda s: (lambda n: hc_sideways(n, max_steps=s, max_sideways=50, time_limit=LIMIT)), [50, 100, 200, 500, 1000])
tune("B. max_steps (1s time limit)",
     lambda s: (lambda n: hc_restart(n, max_steps=s, time_limit=1.0)), [10, 50, 100, 500], ns=(20, 50, 80), seeds=3)
tune("B. time limit in seconds (max_steps=100)",
     lambda s: (lambda n: hc_restart(n, max_steps=100, time_limit=s)), [0.25, 0.5, 1.0, 2.0], ns=(20, 50, 80), seeds=3)
tune("FC. max_backtracks before restart (None = never restart)",
     lambda s: (lambda n: solve_queens_fc(n, max_backtracks=s, time_limit=LIMIT)), [None, 10, 50, 200])
random.seed(42)
```

## Write-up

### How search time was cut down
1. **O(1) conflict counting (all hill-climbing variants).** `ConflictBoard` keeps queen counts per column and per diagonal, so the number of attackers on any square is three array lookups instead of a loop over all queens. Moving a queen updates six counters. Without this, each hill-climbing step at n=100 would cost about n² = 10,000 checks instead of about n = 100.
2. **Random tie-breaking.** When several queens are equally "most constrained", or several columns are equally good, one is picked at random. Without it, A–C keep choosing the same queen and column and fall into cycles.
3. **Sideways moves (C).** Most of the time hill climbing stops on a *plateau*: no single move improves the board, but equal-cost moves exist. Allowing a limited number of sideways moves lets the search walk across the plateau instead of throwing the whole board away.
4. **Cheap queen choice (D).** D picks a random conflicted queen in O(1) instead of scanning all n queens to find the worst one, and it takes the *first* improving column (columns tried in random order) instead of scanning all n columns for the best. Each step costs much less, so D does far fewer conflict checks in total.
5. **Forward checking with bitmask domains and MRV.** Every unassigned row keeps a bitmask of its legal columns. Placing a queen removes its column and diagonals from all future rows at once, and an empty domain triggers an immediate backtrack, so dead ends are found as soon as they're created instead of several rows later. MRV fills the most constrained row first, so the hardest decisions are made while there's still the most freedom. That cut forward checking to about 300 iterations per problem on average (29,488 iterations for all 93 problems), compared with about 3.2 million per problem for DFS on n=8..25.
6. **Validation and time limits.** `run_benchmark` checks every returned board with `is_valid_solution`, and the restarting algorithms take a `time_limit`, so a stuck run counts as "not solved" instead of hanging the notebook.

### How the parameters were chosen (see the tuning output above)
* **D: `max_steps = 1000`.** With 10–100 steps before a restart, D usually restarts before it converges (0–10 of 15 solved). From 500 steps up, all 15 are solved in about 6 ms, and 500, 1000 and 5000 perform the same. 1000 leaves a margin for n=100 without wasting time.
* **C: `max_sideways = 100`, `max_steps = 200`.** With 0 sideways moves (the same as B), only 5 of 15 solved. 10 gave 11 of 15, and from 50 up all were solved. 100 and 200 were the fastest (about 3 ms), so 100 was chosen as the smaller of the two. For `max_steps`, 50 was too few (11 of 15); 100 through 1000 all solved everything with similar times, so 200 was kept.
* **B: `max_steps = 100`, `time_limit = 1s` per n.** Changing `max_steps` made no difference (3 of 9 every time), because B almost never uses up its step budget. It gets stuck (the most constrained queen has no improving move) after a few dozen steps and restarts for that reason. So the *time* limit is the parameter that matters. Going from 0.5s to 2s didn't solve any more of the tuning problems, but it would have doubled the final loop's running time. 1s per n keeps the whole B row near 90 seconds at most and lets it solve the smaller boards.
* **FC: no restarts (`max_backtracks = None`).** With MRV, forward checking solved every tuning problem without restarts in about 8 ms. Restart limits of 10–200 backtracks made little difference, so the simpler version without restarts is used.

### Reflection
**How did the algorithms compare?** British Museum works only for tiny boards: n=8 needed hundreds of thousands of random boards. DFS solves n=8..25, but its time jumps around (n=22 alone takes about 40 million iterations) and it can't reach 30 within 100,000 iterations. Simple hill climbing (A) solved **0 of 93**: it gets stuck within a few dozen moves every time, as expected. Adding restarts (B) helped only for small n (24 of 93), because for large boards the chance that a random start reaches a solution without ever getting stuck is tiny. Sideways moves (C), the cheap random-queen heuristic (D) and forward checking all solved **all 93 problems in under a second in total**.

**Difference in complexity.** A, B and C spend O(n) per step scanning all queens to find the most constrained one, plus another O(n) scanning all columns for the best one. D skips the first scan (a random conflicted queen is O(1) on average) and usually stops the second scan early at the first improving column. In the final table D needs about half as many iterations as C (or fewer) for the same 93 solutions. Choosing the worst queen does make each of C's steps a little more effective, but the extra scan costs about as much as it saves, so the two end up within a few tenths of a second of each other. The fancy check matters most when there's nothing else to help: A and B use it too, and it doesn't save them from getting stuck. The sideways moves (C) and the randomness (D) are what make the difference, not how carefully the queen is chosen.

**How did forward checking fare?** About the same total time as C and D (all under a second for the 93 problems), but with far fewer iterations (about 30K vs. hundreds of thousands), because pruning and MRV mean almost every column it tries is a good one. Each forward-checking iteration costs more (it updates the domains of every unassigned row), which is why the much lower iteration count doesn't make it much faster in seconds. Forward checking is the "rational" approach: it reasons ahead and almost never backtracks. D is the "human" approach: put everything down, then keep fixing whatever's wrong. Both scale easily to n=100.
